## Try your functions

After you change `showdown.py`, restart the kernel (**Restart** in the notebook's toolbar) and run the cells again, so the notebook imports your newest code.

```python
from sklearn.model_selection import StratifiedKFold
from sklearn.tree import DecisionTreeClassifier

from showdown import build_pipeline, compare_models, final_report, make_search, split_bookings, tree_grid

X_train, X_test, y_train, y_test = split_bookings(bookings, 42)
folds = StratifiedKFold(n_splits=5, shuffle=True, random_state=0)
tree_grid([2, 4], [1, 10])
```

```python
search = make_search(build_pipeline(DecisionTreeClassifier(random_state=0)), tree_grid([2, 3, 4, 5, 6, 8, None], [1, 10, 30]), folds)
search.fit(X_train, y_train)
search.best_params_, search.best_score_
```

```python
results = pd.DataFrame(search.cv_results_)
results[["params", "mean_test_score", "std_test_score", "rank_test_score"]].sort_values("rank_test_score").head(8)
```

```python
from sklearn.dummy import DummyClassifier

compare_models({"baseline": build_pipeline(DummyClassifier(strategy="most_frequent")), "tuned tree": search.best_estimator_}, X_train, y_train, folds)
```

## Last of all

Run `final_report(search, X_test, y_test)` once, when your choice is made: the test bookings are for the final score only.
